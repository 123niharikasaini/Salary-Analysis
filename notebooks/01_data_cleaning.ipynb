##### Removing Garbage values

```python
# Filtered to the 5th-99th percentile range based on percentile-jump analysis
# — values below $35K and above $650K showed disproportionate jumps versus 
# adjacent percentiles, indicating likely data entry errors rather than genuine salaries.

df_clean=data[(data['ConvertedCompYearly']>=35000.00) & 
              (data['ConvertedCompYearly']<=650000.00)]
```

Separating the top languages for new binary columns

```python
df_clean['LanguageHaveWorkedWith'].iloc[0]
```

```python
# all language in a list
all_languages = df_clean['LanguageHaveWorkedWith'].dropna().str.split(';').explode()
# frequency for every language
lang_counts = all_languages.value_counts()
```

```python
lang_counts
```

We will take only top 5% languages (frequency>=233) => approx 21 languages

```python
threshold = 0.05 * len(df_clean)
```

```python
# we want only the index values from the lang_counts
top_lang = lang_counts[lang_counts>=threshold].index.tolist()
```

##### Creating new Binary Language Columns

```python
for lang in top_lang:
    df_clean[f'lang_{lang}'] = df_clean['LanguageHaveWorkedWith'].apply(
    lambda x: 1 if lang in x.split(';') else 0
    )
```

```python
def other_lang(x):
    language = x.split(';')
    return 1 if any(lang not in top_lang for lang in language) else 0


df_clean['lang_others'] = df_clean['LanguageHaveWorkedWith'].apply(other_lang)
```

Analysing NULL values (if they are same as `I don't know category` or values left from filling)
* if similar stats merge null values in `I don't know` category

```python
nulls = df_clean[df_clean['OrgSize'].isnull()]

nulls['ConvertedCompYearly'].describe()
```

```python
df_temp = df_clean[df_clean['OrgSize'] == 'I don’t know']
df_temp['ConvertedCompYearly'].describe()
```

```python
from scipy import stats

null_salary = df_clean[df_clean['OrgSize'].isnull()]['ConvertedCompYearly']
dontknow_salary = df_clean[df_clean['OrgSize'] == 'I don’t know']['ConvertedCompYearly']

t_stat, p_value = stats.ttest_ind(null_salary, dontknow_salary, equal_var=False)
print(f"t-statistic: {t_stat:.3f}, p-value: {p_value:.4f}")
```

T-test
* p_value > 0.05 => same groups
* p_value < 0.05 => different groups

p-value > 0.05 => two different population (well our p-value is approx 0.05 but we have huge difference between both mean as well) => Assign New group

The null values doesn't belongs to 'I don’t know'category. That's why imputing the null values with a unknown group.

```python
df_clean['OrgSize'] = df_clean['OrgSize'].fillna('Unknown')
```
